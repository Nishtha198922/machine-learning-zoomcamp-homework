```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv')
```

```python
#Record count

q2_records_count = len(df)
print(f"Question 2 (Records count): {q2_records_count}")
```

```python
#Fuel Types

q3_fuel_types = df['fuel_type'].nunique()
print(f"Question 3 (Fuel types count): {q3_fuel_types}")
```

```python
#Missing Values

q4_columns_with_nan = (df.isnull().sum() > 0).sum()
print(f"Question 4 (Columns with missing values): {q4_columns_with_nan}")
```

```python
#Max Fuel Efficiency

q5_max_fuel_efficiency = df['fuel_efficiency_mpg'].max()
print(f"Question 5 (Max fuel efficiency): {q5_max_fuel_efficiency}")
```

```python
#Median Value of Horsepower

# 1. Calculate the initial median
median_initial = df['horsepower'].median()

# 2. Calculate the mean
mean_val = df['horsepower'].mean()

# 3. Fill missing values with the mean and recalculate median
median_after = df['horsepower'].fillna(mean_val).median()

print(f"Before: {median_initial}, After: {median_after}")
```

```python
#Sum of Weights (Linear Regression Normal Equation)

# 1. Select all the cars from Asia
df_asia = df[df["origin"] == "Asia"]

# 2. Select only columns vehicle_weight and model_year
df_selected = df_asia[["vehicle_weight", "model_year"]]

# 3. Select the first 7 values
df_first_7 = df_selected.head(7)

# 4. Get the underlying NumPy array X
X = df_first_7.values

# 5. Compute matrix-matrix multiplication between X.T and X
XTX = X.T.dot(X)

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)

# 7. Create array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Multiply inverse of XTX with transpose of X, then by y
w = XTX_inv.dot(X.T).dot(y)

# 9. Calculate the sum of all elements in w
sum_of_weights = w.sum()

print(f"Weights w: {w}")
print(f"Sum of weights: {sum_of_weights:.4f}")
```
