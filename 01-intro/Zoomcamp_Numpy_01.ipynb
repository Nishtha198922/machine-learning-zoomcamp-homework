**Creating Arrays**

```python
#importing numpy
import numpy as np
```

```python
#creating an array of zeroes
np.zeros(12)
```

```python
#creating an array of ones
np.ones(14)
```

```python
#creating an array of any number
np.full(11, 100)
```

```python
#making an array from python list
a = np.array([12, 45, 64, 21, 78, 45])
print(a)
```

```python
#to access an element from list
a[3]
```

```python
#to change an element in list 
a[2] = 88
```

```python
a[2]
```

```python
#to create a range of a number
np.arange(40, 51)
```

```python
#to create an array filled with numbers and elenly spaced between first and second parameter
np.linspace(100, 900, 20)
```

**Multi-dimensional arrays**

```python
#to create multi-dimensional array of zeroes
np.zeros((4, 6))
```

```python
#to create multi-dimensional array from list
b = np.array([
    [5, 7, 8, 9],
    [7, 5, 2, 1],
    [9, 3, 0, 6]
])
print(b)
```

```python
#indexing in multi-dimensional array
b[1, 2]
```

```python
#to get the whole row
b[1]
```

```python
#to rewrite the entire row
b[2] = [9, 5, 2, 6]
b[2]
```

```python
#to get the entire column
b[:, 0]
```

```python
#to rewrite the entire column
b[:, 0] = [4, 4, 4]
b[:, 0]
```

**Summarizing operations**

```python
#creating an array
np.random.seed(2)
k = np.random.randint(low=10, high=1000, size=(1,20))
print(k)
```

```python
#to find minimum from the array k
k.min()
```

```python
#to find maximum from array k
k.max()
```

```python
#to find sum of array k
k.sum()
```

```python
#to find standard deviation of array k
k.std()
```

```python
#to find mean of array k
k.mean()
```
