```python
#importing library
import numpy as np
import pandas as pd
```

**DataFrames**

```python
#lists of data
data = [
    ['Nissan', 'Stanza', 1991, 138, 4, 'MANUAL', 'sedan', 2000],
    ['Hyundai', 'Sonata', 2017, None, 4, 'AUTOMATIC', 'Sedan', 27150],
    ['Lotus', 'Elise', 2010, 218, 4, 'MANUAL', 'convertible', 54990],
    ['GMC', 'Acadia',  2017, 194, 4, 'AUTOMATIC', '4dr SUV', 34450],
    ['Nissan', 'Frontier', 2017, 261, 6, 'MANUAL', 'Pickup', 32340],
]

columns = [
    'Make', 'Model', 'Year', 'Engine HP', 'Engine Cylinders',
    'Transmission Type', 'Vehicle_Style', 'MSRP'
]
```

```python
#to access a DataFrame
df = pd.DataFrame(data, columns=columns)
print(df)
```

```python
#to access first two rows
df.head(2)
```

**Series**

```python
#to access a particular column by dot notation
df.Year
```

```python
#to access a particular column using square brackets
df['Transmission Type']
```

```python
#to access multiple columns
df[['Year', 'MSRP', 'Engine Cylinders']]
```

```python
#to add a column
df['id'] = [1, 2, 3, 4, 5]
df
```

```python
#to delete a column
del df['id']
df
```

**Index**

```python
#to access indexes
df.index
```

```python
#access element by there location
df.loc[2]
```

```python
#to change the indexing
df.index = ['a', 'b', 'c', 'd', 'e']
df
```

```python
#to access the element by their index
df.iloc[[0, 3]]
```

```python
#to reset the index
df = df.reset_index(drop=True)
df
```

**Element wise operation**

```python
#multiplication
df['Year'] * 2
```

```python
#comparision
df['Year'] >= 2025
```

**Filtering**

```python
df[
    df['Year'] >= 2010
]
```

```python
#to combine conditions
df[
    (df['Year'] >= 2010) & (df['Make'] == 'Nissan')
]
```

**String**

```python
#to make every word in lowercase
df['Vehicle_Style'].str.lower()
```

```python
#replacing all spaces with underscore
df['Vehicle_Style'] = df['Vehicle_Style'].str.replace(' ', '_').str.lower()
df
```

**Summarizing operators**

```python
df.MSRP.mean()
```

```python
df.MSRP.max()
```

```python
#to get all the statitics at once
df.describe()
```

```python
#to get number of unique values
df.nunique()
```

```python
#to get unique values in a specific column
df.Model.nunique()
```

**Missing Values**

```python
#to find number of missing values
df.isnull().sum()
```

**GroupBy**

```python
df.groupby('Transmission Type').MSRP.max()
```
